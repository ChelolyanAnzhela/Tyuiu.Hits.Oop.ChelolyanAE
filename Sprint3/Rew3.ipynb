```
using System;
using System.Collections.Generic;
public interface IInstallable
{
  void Install();
  
}
public abstract class Component //компоненты
{
    public string Name{get;set;}
    public int Price {get;set;}
    public Component(string name,int price)
    {
        Name=name;
        Price=price;
    }
    public virtual void GetInfo()
    {
     Console.WriteLine($"Компонент:{Name}. Цена: {Price}");
    }
    public  string ToString(string Name)
    {
      return $"Компонент:{Name}. Цена: {Price}";
    }
    
}
public class CPU :Component,IInstallable //процессор
{
  public double Frequency {get;set;}
  public CPU(string name,int price,double frequency) : base(name,price)
  {
    Frequency=frequency;
  }
  public override void GetInfo()
  {
    base.GetInfo();
    Console.WriteLine($"Частота процессора: {Frequency}");
  }
  public void Install()
  {
    Console.WriteLine($"Установка процессора {Name} в сокет");
  }
  public  string ToString(double Frequency)
    {
      return $"Компонент:{Name}. Цена: {Price}Частота процессора: {Frequency}";
    }
}
public class RAM:Component,IInstallable // Оперативная память
{
  public int Volume {get;set;}
  public double Frequency {get;set;}
  public RAM(string name,int price,double frequency, int volume) : base(name,price)
  {
    Volume=volume;
    Frequency=frequency;
  }
  public override void GetInfo()
  {
    base.GetInfo();
    Console.WriteLine($"{Volume}{Frequency}");
  }
  public void Install()
  {
    Console.WriteLine($"Установка оперативной памяти {Name} в слот DIMM");
  }
  public  string ToString(int Volume)
    {
      return $"{Name}{Price}{Volume}";
    }
}
public class GPU:Component,IInstallable // ВИдеокарта
{
  public string Type{get;set;}
  public int Memory{get;set;}
  public GPU(string name,int price,string type,int memory): base(name,price)
  {
    Type=type;
    Memory=memory;
  }
  public override void GetInfo()
  {
    base.GetInfo();
    Console.WriteLine($"{Type}{Memory}");
  }
  public void Install()
  {
   Console.WriteLine($"Установка видеокарты {Name} в слот PCI-Express");
  }
  public string ToString(int Memory)
    {
      return $"{Name}{Price}";
    }
}
public class HDD:Component,IInstallable // жесткий диск
{
 public string Interfaces {get;set;}
 public int Capacity {get;set;}
 public HDD (string name,int price,string interfaces,int capacity) :base(name,price)
 {
    Interfaces=interfaces;
    Capacity=capacity;
 }
 public override void GetInfo()
 {
    base.GetInfo();
    Console.WriteLine($"{Interfaces}{Capacity}");
 }
 public void Install()
 {
    Console.WriteLine($"Установка жесткого диска {Name}");
 }
 public string ToString(int Capacity)
    {
      return $"{Name}{Price}{Capacity}";
    }
}
public class ComputerCollection<T> where T:Component
{
    private List<T> _components = new List<T>();
    public void Add(T components)
    {
     _components.Add(components);
    }
    public void Remove (T components)
    {
        _components.Remove(components);
    }
    public void AllComponents()
    {
        foreach (var components in _components)
        {
            components.GetInfo();
        }
    }
}
ComputerCollection<Component> computer=new ComputerCollection<Component>();
CPU cpu = new CPU("Процессор Intel",2000,25);
cpu.GetInfo();
RAM ram = new RAM ("Оперативная память",2000,25,24);
ram.GetInfo();
GPU gpu = new GPU("ВИдеокарта NVIDIA",2500,"GDDR6 X",32);
gpu.GetInfo();
HDD hdd = new HDD("Жесткий диск Samsung",3000,"NVME",2000);
hdd.GetInfo();
computer.Add(cpu);
computer.Add(ram);
computer.Add(gpu);
computer.Add(hdd);
Console.WriteLine("-----------");
Console.WriteLine("Все компоненты:");
computer.AllComponents();


    
```
