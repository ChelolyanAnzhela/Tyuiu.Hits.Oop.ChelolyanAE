```polyglot-notebook
using System;

public interface IReportable // отчет
{
    void GenerateReport();
}
public interface ITeamLeader // команда
{
    void GetTeamMembers();
}
public class Emploee
{
    public string Name { get; set; } // имя
    public string Position { get; set; } // должность
    public double Salary { get; set; } // зарплата
    public Emploee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }
    public virtual void PrintInfo() // информация сотрудника
    {
        Console.WriteLine($"\nИмя сотрудника:{Name}.\nДолжность сотрудника:{Position}.\nЗарплата сотрудника:{Salary}.");
    }
}
public class Manager : Emploee
{
    protected string Departament { get; set; } // отдел
    public Manager(string name, string position, double salary, string departament) : base(name, position, salary)
    {
        Departament = departament;
    }
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Название управляемого отдела:{Departament}");
    }
}
public class ProjectManager : Manager,IReportable,ITeamLeader
{
    private string ProjectName { get; set; } // название проекта
    public ProjectManager(string name, string position, double salary, string departament, string projectName) : base(name, position, salary, departament)
    {
        ProjectName = projectName;
    }
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект за который отвечает:{ProjectName}");
    }
    public void GenerateReport()
    {
        Console.WriteLine($"Отчет по проекту {ProjectName}:выполнен. Никаких замечаний нет.");
    }
    public void GetTeamMembers()
    {
        
        Console.WriteLine($"Список команды проекта:Евгений Игоревич,Ирина Викторовна,Петр Валентинович");
    }
}
Emploee emploee = new Emploee ("Евгений Игоревич","Стажер",18.500);
emploee.PrintInfo();
Manager manager= new Manager ("Ирина Викторовна","Менеджер",25.000,"Отдел продаж");
manager.PrintInfo();
ProjectManager projname= new ProjectManager("Петр Валентинович","Проектный менеджер",22.500,"Отдел продаж","Проектировка кухни");
projname.PrintInfo();
projname.GenerateReport();
projname.GetTeamMembers()
```
