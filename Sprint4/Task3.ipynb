<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Notification в C#, который будет представлять уведомления
пользователям. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Notification:
• Атрибуты: ID уведомления (NotificationId), Текст уведомления (MessageText),
Тип уведомления (Type).
• Методы:
o DisplayNotification(): метод для отображения уведомления
пользователю.
o SendNotification(): метод для отправки уведомления.
o GetNotificationDetails(): метод для получения деталей уведомления.
Требования к производным классам:
1. EmailУведомление (EmailNotification): Должно содержать дополнительные
атрибуты, такие как Адрес электронной почты (EmailAddress).
Метод SendNotification() должен быть переопределен для отправки
уведомления по электронной почте.
2. SMSУведомление (SMSNotification): Должно содержать дополнительные
атрибуты, такие как Номер телефона (PhoneNumber).
Метод SendNotification() должен быть переопределен для отправки
уведомления через SMS.
3. PushУведомление (PushNotification) (если требуется третий класс): Должно
содержать дополнительные атрибуты, такие как Платформа (Platform,
например, iOS или Android). Метод DisplayNotification() должен быть
переопределен для отображения уведомления на мобильной платформе.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями

```
using System;
using System.Collections.Generic;
public interface IData // интерфейс даты
{
    void Data();
}
public interface IDelivered // интерфейс доставки
{
    void Delivered(bool Deliver);
}
public class Notification
{
    public string Notificationld {get; set;} // атрибут ID
    protected string MessegeText {get; set;} // атрибут сообщения
    public string Type {get; set;} // атрибут типа
    public bool Read {get;set;} // новый атрибут "прочитано"
    protected string Sender {get;set;} // новый атрибут отправитель
    public Notification( string id, string messege, string type,bool read,string sender) // конструктор
    {
        Notificationld=id;
        MessegeText= messege;
        Type=type;
        Read=read;
        Sender=sender;

    }
    public  virtual void DisplayNotification() // отображение
    {
       Console.WriteLine($"\nУведомление {Notificationld}: {MessegeText}");
    }
    public virtual void SendNotification() // отправка
    {
        Console.WriteLine($"Тип уведомления: {Type}"); 
    }
    public virtual string GetNotificationDetails() // детали
    {
        return ($"ID уведомления :{Notificationld}.\nТип уведомления:{Type}.\nСообщение: {MessegeText}.");
    }
    //новые методы
    public virtual void IsRead() // прочитано
    {
        Console.WriteLine($"Уведомление прочитано : {Read}");
    }
    public virtual void Shipper() // отправитель
    {
        Console.WriteLine($"Уведомление {Notificationld} было отправлено отправителем :{Sender}");
    }
}
public class EmailNotification: Notification, IData, IDelivered
{
    private string EmailAdress{get;set;}
    private string Topic {get;set;} // атрибут тема
    public bool Deliver {get;set;} //атрибут доставлен
    public EmailNotification(string id, string messege, string email,bool read,string sender,string topic,bool deliver) : base(id,messege,email,read,sender)
    {
        EmailAdress=email;
        Topic=topic;
        Deliver = deliver;
    }
    public override void SendNotification()
    {
        Console.WriteLine($"Отправленно Электронное уведомление: {Notificationld}.\nНа адресс: {EmailAdress}.| Тема: {Topic}");
    }
    public override string GetNotificationDetails()
    {
        return $"{base.GetNotificationDetails()} | E-mail: {EmailAdress}";
    }
     void IData.Data() //явная реализация интерфейса
    {
        Console.WriteLine($"Электронное уведомление {Notificationld} отправленно 25.10.2025г.");
    }
    public void CallData()// метод для вызова интерфейса
    {
        ((IData)this).Data();
    }
      void IDelivered.Delivered(bool Deliver) //Явная реализация интерфейса
    {
        Console.WriteLine($"Электронное уведомление {Notificationld} было доставлено: {Deliver}");
    }
    public void CallDelivered(bool deliver) // метод для вызова интерфейса
    {
        ((IDelivered)this).Delivered(deliver);
    }
    public override void Shipper() //перекрытие
    {
        Console.WriteLine($"Электронное уведомление {Notificationld} было отправленно отправителем {Sender}");
    }
}
public class SMSNotification : Notification, IData
{
    private string PhoneNumber{get; set;}
    private string Operator{get;set;} // атрибут оператор
    public string Deliver {get;set;}
    public SMSNotification(string id, string messege, string sms,bool read,string sender,string oper, string deliver): base(id,messege,sms,read,sender)
    {
        PhoneNumber=sms;
        Operator=oper;
        Deliver=deliver;
    }
    public override void SendNotification()
    {
      Console.WriteLine($"Отправленно SMS {Notificationld} на номер: {PhoneNumber}");
    }
    public override string GetNotificationDetails()
    {
        return $"{base.GetNotificationDetails()} | Телефон: {PhoneNumber}";
    }
    void IData.Data() //Явная реализация интерфейса
    {
        Console.WriteLine($"SMS уведомление {Notificationld} отправленно 25.10.2025г.");
    }
    public void CallData() // метод для вызова интерфейса
    {
        ((IData)this).Data();
    }
    public void Delivered(string Deliver) // перегрузка
    {
        Console.WriteLine($"SMS уведомление {Notificationld} было доставлено: {Deliver}");
    }
    public override void Shipper() // перекрытие
    {
        Console.WriteLine($"SMS {Notificationld} было отправленно отправителем '{Sender}'. Оператором: {Operator}");
    }
    
}
public class PushNotification : Notification, IData
{
    private string Platform{get; set;}
    public PushNotification(string id, string messege, string platform,bool read,string sender): base(id,messege,platform,read,sender)
    {
        Platform=platform;
    }
    public override void SendNotification()
    {
        Console.WriteLine($"Отправленно push-уведомление: {Notificationld} на устройство {Platform}");
    }
    public override string GetNotificationDetails()
    {
        return $"{base.GetNotificationDetails()} | Платформа: {Platform}";
    }
     void IData.Data()//Явная реализация интерфейса
    {
        Console.WriteLine($"Электронное уведомление {Notificationld} отправленно 25.10.2025г.");
    }
    public void CallData()// метод для вызова интерфейса
    {
        ((IData)this).Data();
    }
    public void Delivered()
    {
        Console.WriteLine($"Электронное уведомление {Notificationld} было доставлено.");
    }
    public override void Shipper()
    {
        Console.WriteLine($"push-уведомление {Notificationld} было отправленно отправителем '{Sender}'");
    }
}
public class NotificationCollection<T> where T: Notification
{
    private List<T> _notification = new List<T>();
    public void Add(T notification)
    {
        _notification.Add(notification);
    }
    public void Remove (T notification)
    {
        _notification.Remove(notification);
    }
    public void Info()
    {
        foreach (var notification in _notification)
        {
            Console.WriteLine(notification.GetNotificationDetails());
        }
    }
}
NotificationCollection<Notification> notification = new NotificationCollection<Notification>();
// создание обьектов и их параметров
EmailNotification emailn=new EmailNotification("Wildberries","В вашей корзине товар по скидке!","@nushe.ru",false,"cats@mail.ru","Добавленный товар",true);
emailn.DisplayNotification();
emailn.SendNotification();

emailn.IsRead();
emailn.Shipper();
emailn.CallData();//вызов интерфейса
emailn.CallDelivered(true);//вызов интерфейса
Console.WriteLine("----------------------");
SMSNotification smsn=new SMSNotification("Телефон","Сделай домашнее задание !","89829467266",true,"Папа","МТС","доставлено");
smsn.DisplayNotification();
smsn.SendNotification();

smsn.IsRead();
smsn.CallData();//вызов интерфейса
smsn.Delivered("доставлено");
smsn.Shipper();
Console.WriteLine("----------------------");
PushNotification push=new PushNotification("Настройки","Новое обновление","IOS",false,"Настройки телефона");
push.DisplayNotification();
push.SendNotification();

push.IsRead();
push.CallData(); //вызов интерфейса
push.Shipper();
Console.WriteLine("----------------------");
Console.WriteLine("КРАТКАЯ ИНФОРМАЦИЯ ВСЕХ УВЕДОМЛЕНИЙ");
Console.WriteLine("----------------------");
notification.Add(emailn);

notification.Add(smsn);

notification.Add(push);
notification.Info();
Console.WriteLine("----------------------");
```
